# 1. ¿Qué es una API?

Una **API (Interfaz de Programación de Aplicaciones)** permite que dos aplicaciones se comuniquen.

Por ejemplo:
- Una aplicación solicita información del clima.
- Un servidor recibe la solicitud.
- El servidor procesa la petición.
- La API devuelve una respuesta que nuestra aplicación puede utilizar.

## Métodos HTTP principales

| Método | ¿Para qué sirve? | Ejemplo |
|---|---|---|
| `GET` | Consultar información | Obtener una lista de usuarios |
| `POST` | Enviar o crear información | Registrar un usuario |
| `PUT` | Actualizar información | Modificar los datos de un usuario |
| `DELETE` | Eliminar información | Borrar un usuario |


## 1.1 ¿Qué es JSON?

Muchas APIs envían sus respuestas en formato **JSON**. JSON organiza la información mediante pares de clave y valor, y se parece a los diccionarios de Python.

Ejemplo de JSON:

```json
{
  "nombre": "Ana",
  "edad": 22,
  "curso": "Python"
}
```

En Python podemos representar esa información con un diccionario:

```python
alumno = {
    "nombre": "Ana",
    "edad": 22,
    "curso": "Python"
}
```

**Recuerda:** JSON es un formato de intercambio de datos; no es lo mismo que un diccionario, aunque su estructura es muy parecida.

## 2.2 Realizar una solicitud GET

Usaremos JSONPlaceholder, una API pública de prueba. Consultaremos una lista de usuarios.

- `requests.get(url)` realiza una solicitud GET.
- `status_code` indica el estado de la respuesta.
- `json()` convierte el contenido JSON de la respuesta en estructuras de datos de Python.


````python
#crear variable URL -> La que vamos consumir
url = "https://jsonplaceholder.typicode.com/users"

# Variable respuesta donde guarderamos lo que hay en nuestra URL
respuesta = requests.get(url)

# status code -> el codigo de nuestra peticion
print("Código de estado:", respuesta.status_code)

# COnvertir el contenido de la respuesta en un json
usuarios = respuesta.json()
# Imprimiendo la respuesta
print(usuarios)
````

### ¿Cómo interpretamos el código de estado?

- `200`: la solicitud se realizó correctamente.
- `404`: no se encontró el recurso solicitado.
- `500`: ocurrió un error en el servidor.

Si recibimos una lista de usuarios, podemos recorrerla como cualquier lista de Python.


````python
for usuario in usuarios:
    print("Nombre:", usuario["name"])
    print("Username:", usuario["username"])
    print("Correo:", usuario["email"])
    print("----------------------")
````

## 2.3 Consultar un recurso específico

También podemos consultar un solo recurso indicando su identificador en la URL.


````python
respuesta = requests.get("https://jsonplaceholder.typicode.com/users/1")

usuario = respuesta.json()

print("Nombre:", usuario["name"])
print("Usuario:", usuario["username"])
print("Correo:", usuario["email"])
````

## 2.4 Enviar información con POST

El método POST permite enviar información al servidor. En este ejemplo, enviaremos un usuario de prueba.

`json=...` indica que enviaremos los datos en formato JSON.


````python
nuevo_usuario = {
    "name": "Val",
    "email": "val@example.com"
}

respuesta = requests.post(
    "https://jsonplaceholder.typicode.com/users",
    json=nuevo_usuario
)

print("Código de estado:", respuesta.status_code)
print("Respuesta:", respuesta.json())
````

## 3.3 Crear un cursor y ejecutar una consulta

El **cursor** permite ejecutar instrucciones SQL desde Python.

- `execute()` ejecuta una instrucción SQL.
- `fetchall()` recupera todas las filas del resultado.
- `fetchone()` recupera una sola fila.


````python
cursor = conexion.cursor()

cursor.execute("SELECT * FROM Productos")

resultados = cursor.fetchall()

print(resultados)

for producto in resultados:
    print(producto)
````

## 3.4 Insertar un registro

Para insertar información usamos `INSERT INTO`. Los valores se envían por separado mediante parámetros `%s`.

Esto ayuda a evitar construir consultas concatenando directamente los datos ingresados.


````python
sql = '''
INSERT INTO alumnos (nombre, edad)
VALUES (%s, %s)
'''

datos = ("Ana", 22)

cursor.execute(sql, datos)
conexion.commit()

print("Registro insertado")
````

# 4. Interacción con MongoDB desde Python

MongoDB es una base de datos orientada a documentos. En lugar de organizar los registros en tablas, utiliza **colecciones** y **documentos**.

Un documento puede tener una estructura similar a JSON.

## 4.1 Comparación de conceptos

| MySQL | MongoDB |
|---|---|
| Base de datos | Base de datos |
| Tabla | Colección |
| Fila | Documento |
| Columna | Campo |

Para trabajar con MongoDB desde Python utilizaremos la biblioteca `pymongo`.


````python
!pip install pymongo
````

````python
from pymongo import MongoClient
````

## 4.2 Conectarse a MongoDB

Podemos conectarnos a una instancia local o a MongoDB Atlas. Aquí utilizaremos una plantilla para Atlas.

Sustituye el URI por el que proporciona tu clúster. Mantén tus credenciales privadas.


````python
uri = "mongodb+srv://user:pass@cluster0.wmnmnoa.mongodb.net/"

cliente = MongoClient(uri)

db = cliente["escuela"]
coleccion = db["alumnos"]

print("Conexión configurada")
````

## 4.3 Insertar un documento

`insert_one()` inserta un documento en la colección.


````python
alumno = {
    "nombre": "Ana",
    "edad": 22,
    "curso": "Python"
}

resultado = coleccion.insert_one(alumno)

print("ID del documento:", resultado.inserted_id)
````

## 4.4 Consultar documentos

`find()` permite consultar documentos de una colección. Si no indicamos un filtro, devuelve los documentos de la colección.


````python
resultados = coleccion.find()

for alumno in resultados:
    print(alumno)
````

También podemos buscar documentos que cumplan una condición.


````python
resultado = coleccion.find({"curso": "Python"})

for alumno in resultado:
    print(alumno)
````

## 4.5 Actualizar un documento

`update_one()` actualiza el primer documento que coincide con el filtro. El operador `$set` modifica el campo indicado.


````python
coleccion.update_one(
    {"nombre": "Ana"},
    {"$set": {"edad": 23}}
)
````

## 4.6 Eliminar un documento

`delete_one()` elimina el primer documento que coincide con el filtro.

**Precaución:** verifica el filtro antes de ejecutar una eliminación.


````python
coleccion.delete_one({"nombre": "Ana"})
````

# 5. Comparación: MySQL y MongoDB

Ambas tecnologías permiten almacenar y administrar información, pero utilizan estructuras diferentes.

| Operación | MySQL | MongoDB |
|---|---|---|
| Consultar | `SELECT` | `find()` |
| Insertar | `INSERT` | `insert_one()` |
| Actualizar | `UPDATE` | `update_one()` |
| Eliminar | `DELETE` | `delete_one()` |

## Ideas clave de la sesión

- Una API permite que distintas aplicaciones se comuniquen.
- GET consulta información y POST envía información para crear un recurso.
- `requests` permite consumir APIs desde Python.
- `mysql-connector-python` permite conectarnos a MySQL.
- `pymongo` permite conectarnos a MongoDB.
- MySQL organiza datos en tablas y MongoDB en colecciones y documentos.
