Desarrolla un programa en Python que permita a los usuarios registrar y actualizar el estado de varios proyectos. Los proyectos pueden estar en diferentes estados: no iniciado, en progreso y completado. El programa permitirá actualizar y mostrar el estado de los proyectos, y permanecerá en ejecución hasta que el usuario decida salir. 

El programa debe realizar las siguientes acciones:
1. Solicitará al usuario seleccione que una opción: registrar nuevo proyecto, actualizar estado de proyecto, mostrar lista de proyectos o salir.
2. Dependiendo la opción, solicitará al usuario el nombre del proyecto.
3. Pedirá al usuario que ingrese el estado del proyecto (no iniciado, en progreso, completado).
4. Permitirá al usuario actualizar el estado de los proyectos registrados.
5. Mostrará una lista de todos los proyectos con su estado actual.
6. Finalmente, el programa debe mostrar un mensaje con el nombre y el estado de todos los proyectos registrados cuando el usuario decida mostrar la lista de proyectos.


Para ello, el programa debe seguir estos pasos:
- **Registrar el nombre del proyecto.** Solicitar al usuario que ingrese el nombre del proyecto que desea registrar.
- **Solicitar el estado del proyecto.** Pedir al usuario que ingrese el estado actual del proyecto. Los estados pueden ser "no iniciado", "en progreso" o "completado".
- **Actualizar el estado del proyecto.** Permitir que el usuario actualice el estado de un proyecto existente.
- **Mostrar la lista de proyectos.** Utilizar un bucle para iterar a través de la lista de proyectos y mostrar su estado actual.
- **Ejecución continua.** Mantener el programa en ejecución hasta que el usuario decida salir, permitiendo múltiples operaciones.

```python
# Gestor de Proyectos :)
ESTADOS = {
    "1": "No Iniciado",
    "2": "En Progreso",
    "3": "Finalizado"
}

def pedir_estado():
    while True:
        print("\nEstados disponibles:")
        print("1. No Iniciado")
        print("2. En Progreso")
        print("3. Finalizado")

        opcion = input("\nSelecciona el estado de tu proyecto (1-3): ").strip()

        match opcion:
            case "1" | "2" | "3":
                return ESTADOS[opcion]

            case _:
                print("Estado inválido. Inténtalo de nuevo.")

def registrar_proyecto(proyectos):
    nombre_proyecto = input("\nNombre del proyecto: ").strip()

    if nombre_proyecto == "":
        print("El nombre del proyecto no puede estar vacío.")

    elif nombre_proyecto.lower() in proyectos:
        print(f"El proyecto '{nombre_proyecto}' ya existe. Si deseas actualizarlo, usa la opción 2.")

    else:
        estado = pedir_estado()

        proyectos[nombre_proyecto.lower()] = {
            "nombre": nombre_proyecto,
            "estado": estado
        }

        print(f"El proyecto '{nombre_proyecto}' fue registrado correctamente con el estado: '{estado}'.")

def actualizar_proyecto(proyectos):
    if not proyectos:
        print("\nAún no hay proyectos registrados.")
        return

    mostrar_proyectos(proyectos)

    nombre_proyecto = input("\n¿Qué proyecto deseas actualizar? ").strip()

    if nombre_proyecto.lower() in proyectos:
        nuevo_estado = pedir_estado()

        proyectos[nombre_proyecto.lower()]["estado"] = nuevo_estado

        print(f"El proyecto '{nombre_proyecto}' ahora está '{nuevo_estado}'.")

    else:
        print(f"No existe un proyecto con el nombre '{nombre_proyecto}'.")


def mostrar_proyectos(proyectos):
    if not proyectos:
        print("\nAún no hay proyectos registrados.")
        return

    print("\n-----------------------------------------------")
    print("|                 Proyectos                   |")
    print("-----------------------------------------------")

    for numero, proyecto in enumerate(proyectos.values(), start=1):
        print(f"\n{numero}. {proyecto['nombre']} -> {proyecto['estado']}")


def main():
    proyectos = {}
    ejecutando = True

    while ejecutando:
        print("\n-----------------------------------------------")
        print("|      Bienvenidx al Gestor de Proyectos      |")
        print("-----------------------------------------------")

        print("\n1. Registrar un nuevo proyecto.")
        print("2. Actualizar estado de un proyecto.")
        print("3. Ver todos los proyectos.")
        print("4. Salir")

        opcion = input("\nSelecciona una opción (1-4): ").strip()

        match opcion:
            case "1":
                registrar_proyecto(proyectos)

            case "2":
                actualizar_proyecto(proyectos)

            case "3":
                mostrar_proyectos(proyectos)

            case "4":
                total = len(proyectos)
                print(f"\nProyectos registrados: {total}.")
                print("Fin del programa.")
                ejecutando = False

            case _:
                print("Opción inválida. Inténtalo de nuevo.")


if __name__ == "__main__":
    main()
```
