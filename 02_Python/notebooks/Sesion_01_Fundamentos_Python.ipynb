## 2. Tipos de datos básicos

| Tipo | Para qué sirve | Ejemplo |
|---|---|---|
| `int` | Números enteros | `25` |
| `float` | Números con decimales | `8.75` |
| `str` | Texto | `"Python"` |
| `bool` | Verdadero o falso | `True` |
| `list` | Colección modificable | `["SQL", "Python"]` |
| `tuple` | Colección no modificable | `("SQL", "Python")` |
| `dict` | Pares clave-valor | `{"nombre": "Ana"}` |

Nota: las listas, tuplas y diccionarios pueden contener varios valores; más adelante serán especialmente útiles al procesar datos.

````python
cantidad_estudiantes = 25            # int
promedio_grupo = 8.75                # float
nombre_modulo = "Python"             # str
sesion_completada = True             # bool
temas = ["variables", "operadores"]  # list
coordenadas = (19.4, -99.1)          # tuple
estudiante = {"nombre": "Ana", "edad": 19}  # dict

print(type(cantidad_estudiantes))
print(type(promedio_grupo))
print(type(nombre_modulo))
print(type(sesion_completada))
print(type(temas))
print(type(coordenadas))
print(type(estudiante))
````

### Cuidado con las comillas

`25` y `"25"` no son lo mismo:
- `25` es un entero y puede usarse directamente en cálculos.
- `"25"` es texto; para sumarlo como número debemos convertirlo.

````python
edad_numero = 25
edad_texto = "25"

print(type(edad_numero))
print(type(edad_texto))

# Descomenta la siguiente línea para observar el error:
# print(edad_texto + 5)

print(int(edad_texto) + 5)
````

## 3. Operadores aritméticos

| Operador | Significado | Ejemplo |
|---|---|---|
| `+` | Suma | `8 + 3` |
| `-` | Resta | `8 - 3` |
| `*` | Multiplicación | `8 * 3` |
| `/` | División | `8 / 3` |
| `//` | División entera | `8 // 3` |
| `%` | Residuo | `8 % 3` |
| `**` | Potencia | `2 ** 3` |

````python
precio_curso = 1200
descuento = 250
cantidad = 2

print("Precio final por curso:", precio_curso - descuento)
print("Total por dos cursos:", (precio_curso - descuento) * cantidad)
print("17 / 5 =", 17 / 5)
print("17 // 5 =", 17 // 5)
print("17 % 5 =", 17 % 5)
print("2 ** 3 =", 2 ** 3)
````

## 4. Operadores relacionales

Comparan dos valores y devuelven un booleano (`True` o `False`).

- `==` igual a
- `!=` diferente de
- `>` mayor que
- `<` menor que
- `>=` mayor o igual que
- `<=` menor o igual que

**Importante:** `=` asigna un valor; `==` compara dos valores.

````python
edad = 20
minimo_edad = 18

print(edad == 20)
print(edad != 18)
print(edad > minimo_edad)
print(edad < 18)
print(edad >= 20)
print(edad <= 19)
````

## 5. Operadores lógicos

- `and`: ambas condiciones deben ser verdaderas.
- `or`: al menos una condición debe ser verdadera.
- `not`: invierte el valor lógico.

Ejemplo: una persona puede recibir una beca si tiene promedio de al menos 9 **y** asistencia de al menos 90%.

````python
promedio = 9.3
asistencia = 95

puede_recibir_beca = promedio >= 9 and asistencia >= 90
print("¿Puede recibir la beca?", puede_recibir_beca)

tiene_cupon = False
es_estudiante = True

hay_beneficio = tiene_cupon or es_estudiante
print("¿Tiene algún beneficio?", hay_beneficio)

print("¿No es estudiante?", not es_estudiante)
````

## 6. Operadores de asignación

Los operadores compuestos actualizan el valor de una variable:
- `x += 3` equivale a `x = x + 3`
- `x -= 2` equivale a `x = x - 2`
- `x *= 4` equivale a `x = x * 4`
- `x /= 2` equivale a `x = x / 2`
- `x %= 3` guarda el residuo de dividir entre 3

Observa cómo cambia la variable paso a paso.

````python
puntos = 10
print("Inicio:", puntos)

puntos += 5
print("Después de sumar 5:", puntos)

puntos -= 3
print("Después de restar 3:", puntos)

puntos *= 2
print("Después de multiplicar por 2:", puntos)

puntos %= 7
print("Residuo al dividir entre 7:", puntos)
````

## 7. Strings e interpolación

Podemos insertar valores dentro de un texto. En Python moderno, los **f-strings** suelen ser la opción más legible.

También existen `%` y `.format()`, que conviene reconocer porque aparecen en código existente.

````python
nombre = "Mariana"
curso = "Python"
precio = 1200

# F-string: opción recomendada
print(f"Hola, {nombre}. El curso de {curso} cuesta ${precio}.")

# Método format()
print("Hola, {}. El curso de {} cuesta ${}.".format(nombre, curso, precio))

# Formato antiguo con %
print("Hola, %s. El curso de %s cuesta $%d." % (nombre, curso, precio))
````

### Formatear decimales en un f-string

Por ejemplo, `:.2f` muestra dos decimales.

````python
promedio = 9.4567
print(f"Tu promedio es {promedio:.2f}")
````
