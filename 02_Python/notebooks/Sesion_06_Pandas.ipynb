## 2. Series: una columna de datos
Una **Serie** es una estructura unidimensional de Pandas. Tiene valores y un índice que permite identificar cada elemento.

```python
precios = pd.Series([28.00, 34.50, 26.00, 40.00],
                     index=["Aceite", "Arroz", "Azúcar", "Huevo"])

print(precios)
```

### Consultar valores de una Serie
- `loc` consulta por etiqueta.
- `iloc` consulta por posición.

```python
print("Precio del arroz:")
print(precios.loc["Arroz"])

print("Primer precio:")
print(precios.iloc[0])

print("Primeros tres precios:")
print(precios.iloc[:3])
```

## 5. Manipulación de columnas
Podemos agregar, renombrar y eliminar columnas. Estas operaciones modifican la estructura de la tabla.

```python
# Agregar una columna constante
df["moneda"] = "MXN"

# Renombrar una columna
df = df.rename(columns={"precio": "precio_mxn"})

df[["producto", "precio_mxn", "moneda"]].head()
```

```python
# Crear una columna derivada (ejemplo didáctico)
df["precio_doble"] = df["precio_mxn"] * 2

df[["producto", "precio_mxn", "precio_doble"]].head()
```

```python
# Eliminar una columna
df = df.drop(columns=["moneda"])

df.head()
```
