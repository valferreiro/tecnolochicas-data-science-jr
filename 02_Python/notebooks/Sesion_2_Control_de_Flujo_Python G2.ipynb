## 1. ¿Qué es el control de flujo?

Normalmente, Python ejecuta las instrucciones de arriba hacia abajo. Las estructuras de control de flujo permiten cambiar ese comportamiento:

- **Condiciones:** el programa elige qué instrucciones ejecutar.
- **Ciclos:** el programa repite instrucciones.

Por ejemplo, una aplicación puede decidir si una persona aprobó una materia o recorrer todos los productos de una tienda.

```python
print("Inicio del programa")
print("Esta instrucción se ejecuta después")
print("Fin del programa")
```

## 2. Condiciones: `if`

`if` significa “si”. Python evalúa una condición y ejecuta el bloque indentado solamente si el resultado es verdadero (`True`).

Los operadores de comparación más comunes son:

| Operador | Significado |
|---|---|
| `==` | Igual a |
| `!=` | Diferente de |
| `>` | Mayor que |
| `<` | Menor que |
| `>=` | Mayor o igual que |
| `<=` | Menor o igual que |

La indentación —los espacios al inicio de una línea— indica qué instrucciones pertenecen al `if`.

```python
temperatura = 29

# Si temperatura es mayor a 25
if temperatura > 25:
    print("Hace calor")

print("Esta línea se ejecuta siempre")
```

```python
temperatura = int(input("Introduce la temperatura actual: "))
print(temperatura)
```

### ¿Qué sucede en el ejemplo?

1. La variable `temperatura` guarda el valor `29`.
2. Python comprueba si `29 > 25`.
3. Como la condición es verdadera, muestra `Hace calor`.
4. La última instrucción está fuera del bloque `if`, por eso se ejecuta independientemente de la condición.

**Prueba:** cambia `temperatura` a `18` y ejecuta de nuevo. La condición será falsa y no aparecerá el mensaje `Hace calor`.

```python
# Prueba con diferentes valores
temperatura = 18

if temperatura > 25:
    print("Hace calor")
```

## 3. Dos caminos: `if` y `else`

`else` significa “si no”. Se utiliza cuando queremos definir qué hacer si la condición del `if` es falsa.

El programa elige uno de los dos caminos: el bloque de `if` o el bloque de `else`.

```python
# Definimos variable edad
edad = 16

# Si edad 16>18
if edad >= 18:
  #True imprime Puedes registrarte
    print("Puedes registrarte")
#Si no
else:
  #False imprime Necesia autorizacion
    print("Necesitas autorización")
```

### ¿Cómo funciona?

- Si `edad` es mayor o igual a 18, se ejecuta el primer mensaje.
- Si la condición no se cumple, se ejecuta el mensaje de `else`.
- `else` no lleva una condición después de la palabra.

**Prueba:** cambia `edad` a `20` y después a `15`. Observa qué bloque se ejecuta en cada caso.

```python
edad = 20

if edad >= 18:
    print("Puedes registrarte")
else:
    print("Necesitas autorización")
```

## 4. Varias posibilidades: `if`, `elif` y `else`

`elif` significa “si no, si…”. Permite evaluar otra condición cuando la anterior fue falsa.

Python evalúa las condiciones de arriba hacia abajo y ejecuta solamente el primer bloque cuya condición sea verdadera. Si ninguna se cumple, ejecuta `else`.

```python
# Guardar variable calificacion
calificacion = 6

# SI 5 > 6 = False
if calificacion >= 9:
    print("Excelente")
# Si no cumplio la anterioir
# verifica si 5 > 7 = False
elif calificacion >= 7:
    print("Aprobada")
 # Si no cumplio la anterior
 # Verifiva si 5 >= 6 = False
elif calificacion == 6:
    print("Necesita reforzar algunos temas")
else:
    print("Debe repetir la evaluación")
```

En este ejemplo, una calificación de `8` no cumple la primera condición (`>= 9`), pero sí cumple la segunda (`>= 7`). Por eso imprime `Aprobada` y no revisa los bloques siguientes.

**Prueba:** ejecuta el código con `10`, `7`, `6` y `4`. Nota que el orden de las condiciones importa.

```python
calificacion = 6

if calificacion >= 9:
    print("Excelente")
elif calificacion >= 7:
    print("Aprobada")
elif calificacion >= 6:
    print("Necesita reforzar algunos temas")
else:
    print("Debe repetir la evaluación")
```

## 5. Ciclos: ¿por qué repetir instrucciones?

Un ciclo permite ejecutar un bloque de código varias veces sin escribir la misma instrucción repetidamente.

Python tiene distintos ciclos. En esta sesión usaremos `for` y `while`.

## 6. El ciclo `for`

`for` se utiliza para recorrer los elementos de una secuencia, como una lista de productos, una cadena de texto o un rango de números.

En cada repetición, la variable del ciclo toma el siguiente elemento.

```python
print("Producto:", "Laptop")
print("Producto:", "Mouse")
print("Producto:", "Teclado ")
print("Producto:", "Monitor")
```

```python
productos = ["Laptop", "Mouse", "Teclado", "Monitor"]
numeros = [1,2,3,4,5,6,7,8,9,0]

for producto in productos:
    print("Producto:", producto)


print("Producto:", productos[0])
```

### Paso a paso

- `productos` es una lista con cuatro elementos.
- `for producto in productos` recorre la lista.
- En cada vuelta, `producto` guarda un elemento diferente.
- `print()` muestra el elemento actual.
- Cuando no quedan elementos, el ciclo termina.

Este tipo de ciclo es útil cuando queremos procesar todos los elementos de una lista.

### Usar `for` con `range()`

`range()` genera una secuencia de números. Por ejemplo, `range(1, 6)` produce los números desde `1` hasta `5`: el límite final no se incluye.

```python
for numero in range(1, 6):
    print(numero)
```

### `for` con una condición

Podemos combinar un ciclo y una condición. Por ejemplo, mostrar solamente los productos que coinciden con una búsqueda.

```python
productos = ["Laptop", "Mouse", "Teclado", "Monitor"]
busqueda = "Monitor"

for producto in productos:
    if producto == busqueda:
        print("Producto encontrado:", producto)
    print("Finalizando ciclo de busqueda", producto)
else:
    print("Termino la busqueda")
```

### ¿Qué significa `for-else`?

Un ciclo `for` puede tener un bloque `else`. Este bloque se ejecuta cuando el ciclo termina normalmente, es decir, cuando recorrió la secuencia completa sin ser interrumpido por `break`.

Es una característica menos común; primero asegúrate de comprender el `for` básico.

```python
for numero in range(3):
    print("Número:", numero)
else:
    print("El ciclo terminó normalmente")
```

## 7. `match-case`: elegir entre opciones

`match-case` permite comparar un valor con diferentes patrones. Es útil, por ejemplo, para organizar las opciones de un menú.

- `match` indica qué valor se va a revisar.
- Cada `case` define una opción posible.
- `case _` es el caso predeterminado si ninguna de las opciones anteriores coincide.

`match-case` está disponible en Python 3.10 o posterior.

```python
opcion = 1

match opcion:
    case 1:
        print("Consultar saldo")
    case 2:
        print("Realizar pago")
    case 3:
        print("Salir")
    case _:
        print("Opción no válida")
```

Como `opcion` vale `2`, se ejecuta el segundo caso. Cambia el valor a `1`, `3` y `8` para observar los resultados.

### ¿Cuándo usar `match-case` en lugar de `if`?

- `match-case` resulta claro cuando comparamos un valor con opciones concretas.
- `if-elif-else` es más flexible para condiciones como `edad >= 18`, rangos o combinaciones lógicas.

```python
# Ejemplo de comparación con if-elif
opcion = 3

if opcion == 1:
    print("Consultar saldo")
elif opcion == 2:
    print("Realizar pago")
elif opcion == 3:
    print("Salir")
else:
    print("Opción no válida")
```

## 8. El ciclo `while`

`while` repite un bloque de instrucciones **mientras una condición sea verdadera**.

A diferencia de `for`, suele utilizarse cuando la cantidad de repeticiones depende de una condición y no necesariamente se conoce desde el inicio.

Es importante que el ciclo tenga una forma de terminar. Si la condición nunca se vuelve falsa, se puede crear un ciclo infinito.

```python
intentos = 1

while intentos <= 3:
    print("Intento número:", intentos)
    intentos += 1

print("Fin del proceso")
```

### Paso a paso

1. `intentos` comienza en `1`.
2. Python comprueba si `intentos <= 3`.
3. Si la condición es verdadera, imprime el intento.
4. `intentos += 1` aumenta el contador en uno.
5. Cuando `intentos` vale `4`, la condición es falsa y el ciclo termina.

La instrucción `intentos += 1` equivale a `intentos = intentos + 1`.

### ¿Qué es un ciclo infinito?

Si el contador no cambia, la condición puede permanecer verdadera para siempre. Por ejemplo, el siguiente código está comentado para evitar que se ejecute accidentalmente. **No elimines el símbolo `#` de este ejemplo y lo ejecutes sin corregirlo.**

```python
# Ejemplo de ciclo infinito (no ejecutar así):
#
#intentos = 1
#while intentos <= 3:
#     print(intentos)
#
# Falta aumentar intentos, así que siempre vale 1.
```

### `while-else`

El `else` de `while` se ejecuta cuando el ciclo termina porque la condición se volvió falsa. Si el ciclo se interrumpe con `break`, el bloque `else` no se ejecuta.

```python
intentos = 0

while intentos < 2:
    print("Intentando...")
    intentos += 1
else:
    print("La condición dejó de cumplirse")
```

## 9. ¿Qué estructura conviene utilizar?

| Estructura | ¿Para qué sirve? | Ejemplo |
|---|---|---|
| `if` | Ejecutar código si una condición es verdadera. | Revisar si hay existencias. |
| `if-else` | Elegir entre dos caminos. | Determinar si una persona es mayor de edad. |
| `if-elif-else` | Elegir entre varias condiciones. | Clasificar una calificación. |
| `for` | Recorrer una secuencia o repetir un número conocido de veces. | Mostrar los productos de una lista. |
| `match-case` | Elegir entre opciones concretas. | Procesar una opción de menú. |
| `while` | Repetir mientras una condición sea verdadera. | Mantener activo un menú. |

Una misma solución puede combinar varias estructuras. Por ejemplo, un menú puede usar `while` para repetirse y `if` o `match-case` para decidir qué acción realizar.
