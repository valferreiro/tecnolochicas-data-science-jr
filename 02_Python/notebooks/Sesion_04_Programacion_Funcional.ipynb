## 1. Definición de funciones

Una **función** es un bloque de código que realiza una tarea específica. No se ejecuta por el simple hecho de estar escrita: se ejecuta cuando la **llamamos**.

Piensa en una función como una pequeña máquina:
- recibe información (parámetros),
- realiza instrucciones,
- y puede entregar un resultado (`return`).

### Partes de una función
- `def`: palabra reservada para definirla.
- Nombre: identifica la función.
- Parámetros: datos que puede recibir.
- Cuerpo: instrucciones indentadas.
- `return`: devuelve un resultado; es opcional. Si no se usa, Python devuelve `None`.


````python
def saludar(nombre):
    mensaje = "Hola, " + nombre + "!"
    return mensaje

nombre_input = input("Cual es tu nombre? ")
#LLAMANDO FUNCION
resultado = saludar(nombre_input)
print(resultado)
````

**Resultado esperado**
```text
Hola, Ana!
```

**¿Qué ocurrió?**
1. `def saludar(nombre):` creó la función.
2. `"Ana"` se envió como argumento al llamar `saludar("Ana")`.
3. La función construyó el mensaje.
4. `return` entregó el mensaje y lo guardamos en `resultado`.

**Parámetro y argumento no son lo mismo:**
- `nombre` es el parámetro: el espacio que espera un dato.
- `"Ana"` es el argumento: el valor que enviamos al llamar la función.


### Una función puede recibir más de un parámetro

Por ejemplo, una tienda puede calcular el total de una compra multiplicando el precio por la cantidad.


````python
# Definimos mi funcion -> calcular_total
# tiene de parametro 2 variables
# precio
# cantidad
def calcular_total(precio, cantidad):
    total = precio * cantidad
    return total

total_compra = calcular_total(25, 3)
print("Total de la compra:", total_compra)
````

````python
def calcular_total(precio, cantidad):
    total = precio * cantidad
    return total

precio = int(input("¿Cuánto cuesta? "))
cantidad = int(input("¿Cuántos compraste? "))
# total_compra = calcular_total(25, 3)
# print("Total de la compra:", total_compra)
print("Total de la compra:", calcular_total(precio,cantidad))
````

**Resultado esperado**
```text
Total de la compra: 75
```

La función recibe dos datos, realiza una operación y devuelve un único resultado. Podemos reutilizarla con diferentes precios y cantidades sin escribir la multiplicación cada vez.


### `print()` y `return` cumplen funciones diferentes

- `print()` muestra información en pantalla.
- `return` devuelve un valor para que podamos guardarlo, reutilizarlo o realizar otra operación.

Observa la diferencia:


````python
def mostrar_total(precio, cantidad):
    print(precio * cantidad)
    return precio * cantidad

def devolver_total(precio, cantidad):
    return precio * cantidad

resultado_print = mostrar_total(10, 2)
resultado_return = devolver_total(10, 2)

print("Valor obtenido de la función con print:", resultado_print)
print("Valor obtenido de la función con return:", resultado_return)
````

**Resultado esperado**
```text
20
Valor obtenido de la función con print: None
Valor obtenido de la función con return: 20
```

La primera función muestra `20`, pero no devuelve ese valor; por eso la variable recibe `None`. La segunda sí devuelve `20`.


## 2. Funciones lambda

Una función `lambda` es una función anónima, es decir, no necesita escribirse con un nombre usando `def`. Se utiliza normalmente para operaciones pequeñas que pueden expresarse en una sola línea.

### Comparación: función tradicional y lambda


````python
# Función tradicional
def sumar(a, b):
    return a + b

print(sumar(2, 4))

# La misma operación con lambda
suma = lambda a, b: a + b

print(suma(2, 4))
````

**Resultado esperado**
```text
6
6
```

Ambas realizan exactamente la misma operación. En la lambda:
- `lambda` inicia la expresión.
- `a, b` son los parámetros.
- `a + b` es la expresión cuyo resultado se devuelve automáticamente.

**¿Cuándo conviene usarla?** Cuando la operación es breve y sencilla. Si necesitamos varias instrucciones o una lógica compleja, es más claro utilizar `def`.


### Otro ejemplo: calcular un descuento

Supongamos que queremos aplicar un descuento del 10 % a distintos precios.


````python
aplicar_descuento = lambda precio: precio * 0.90

precio_original = 200
precio_final = aplicar_descuento(precio_original)

print("Precio original:", precio_original)
print("Precio con descuento:", precio_final)
````

**Resultado esperado**
```text
Precio original: 200
Precio con descuento: 180.0
```

La lambda recibe el precio y devuelve el 90 % de ese valor. El resultado es `180.0`.


## 3. Control de excepciones

Mientras un programa se ejecuta, pueden aparecer errores. Por ejemplo, intentar dividir entre cero o convertir una palabra en un número.

El control de excepciones permite **anticipar y gestionar** ciertos errores para que el programa responda de forma controlada.

### Bloques principales
- `try`: contiene el código que podría generar un error.
- `except`: indica qué hacer si ocurre una excepción.
- `else`: se ejecuta si el bloque `try` terminó sin errores.
- `finally`: se ejecuta siempre, haya ocurrido un error o no.

No es obligatorio utilizar los cuatro bloques en todos los casos.


### Ejemplo 1: evitar una división entre cero

Primero observa qué sucede cuando el divisor es cero.


````python
#try:
resultado = 10 / 0
print(resultado)
#except ZeroDivisionError:
#    print("No es posible dividir entre cero.")
````

**Resultado esperado**
```text
No es posible dividir entre cero.
```

Python detectó `ZeroDivisionError` dentro de `try` y pasó a ejecutar el bloque `except`. Así mostramos un mensaje entendible en lugar de dejar que el programa termine con un error sin gestionar.


### Ejemplo 2: utilizar `try`, `except` y `else`

En una tienda, el usuario podría escribir la cantidad de productos. Necesitamos convertir ese dato a entero.


````python
try:
    cantidad = int("3")
except ValueError:
    print("Debes escribir una cantidad numérica.")
else:
    print("Cantidad registrada:", cantidad)
````

**Resultado esperado**
```text
Cantidad registrada: 3
```

Como `"3"` pudo convertirse a entero, no se ejecutó `except`; se ejecutó `else`.

Si cambiamos el texto `"3"` por `"tres"`, se producirá un `ValueError` y se ejecutará el mensaje de `except`.


### Ejemplo 3: el bloque `finally`

`finally` se ejecuta siempre. Puede servir para realizar acciones finales, como cerrar un recurso o mostrar que un proceso terminó.


````python
try:
    numero = int("abc")
except ValueError:
    print("El dato ingresado no es un número.")
finally:
    print("Fin del proceso.")
````

**Resultado esperado**
```text
El dato ingresado no es un número.
Fin del proceso.
```

Aunque ocurrió un error y se ejecutó `except`, el bloque `finally` también se ejecutó.


### Elegir una excepción específica

Es recomendable capturar el tipo de error que esperamos manejar. Por ejemplo:
- `ValueError`: un valor no tiene el formato esperado.
- `ZeroDivisionError`: se intenta dividir entre cero.

Esto permite mostrar mensajes más claros y evita ocultar errores que no habíamos previsto.


````python
try:
    edad = int("veinte")
except ValueError:
    print("La edad debe escribirse con números.")
````

**Resultado esperado**
```text
La edad debe escribirse con números.
```


## 4. `map()`, `filter()` y `reduce()`

Estas funciones trabajan con iterables, como listas o tuplas, y permiten aplicar operaciones a sus elementos.

- `map()`: transforma cada elemento.
- `filter()`: conserva los elementos que cumplen una condición.
- `reduce()`: combina los elementos hasta obtener un solo resultado.

Vamos a utilizar una misma lista de precios para entender la diferencia.


### 4.1 `map()`: transformar elementos

`map(funcion, iterable)` aplica una función a cada elemento y devuelve un iterador con los resultados.

Ejemplo: calcular el precio con IVA para cada producto.


````python
precios = [100, 200, 50, 80]
precios_con_iva = map(lambda precio: precio * 1.16, precios)

print(list(precios_con_iva))
````

````python
precios = [100, 200, 50, 80]

def aumentar_precio(precio):
    return precio + 50

precios_aumentados = map(aumentar_precio, precios)

print(list(precios_aumentados))
````

**Resultado esperado**
```text
[116.0, 232.0, 58.0, 92.8]
```

`map()` tomó cada precio, aplicó la lambda y produjo una nueva secuencia de valores. Usamos `list()` para visualizar todos los elementos del iterador.


### 4.2 `filter()`: seleccionar elementos

`filter(funcion, iterable)` conserva únicamente los elementos para los que la función devuelve `True`.

Ejemplo: seleccionar precios mayores o iguales a 100.


````python
precios = [100, 200, 50, 80, 150]

precios_altos = filter(lambda precio: precio >= 100, precios)

print(list(precios_altos))
````

````python
precios = [100, 200, 50, 80, 150]

def filtrar_precios(precio):
    return precio >= 100

precios_altos = filter(filtrar_precios, precios)

print(list(precios_altos))
````

**Resultado esperado**
```text
[100, 200, 150]
```

La condición se evalúa para cada precio. Los que cumplen `precio >= 100` se conservan; los demás se descartan.


### 4.3 `reduce()`: combinar elementos en un solo valor

`reduce()` aplica repetidamente una función para combinar los elementos de un iterable y obtener un único resultado. Se importa desde el módulo `functools`.

Ejemplo: sumar todos los precios.


````python
from functools import reduce

precios = [100, 200, 50, 80]

total = reduce(lambda acumulado, precio: acumulado + precio, precios)

print("Total:", total)
````

````python
from functools import reduce

precios = [100, 200, 50, 80]

def sumar_precios(acumulado, precio):
    return acumulado + precio

total = reduce(sumar_precios, precios)

print("Total:", total)
````

**Resultado esperado**
```text
Total: 430
```

Podemos imaginar el proceso así:
1. Toma 100 y 200: obtiene 300.
2. Combina 300 con 50: obtiene 350.
3. Combina 350 con 80: obtiene 430.

El acumulado va guardando el resultado parcial hasta obtener un solo valor.


### Comparación rápida

| Función | ¿Qué hace? | Ejemplo de uso |
|---|---|---|
| `map()` | Transforma cada elemento | Calcular IVA de cada precio |
| `filter()` | Selecciona según una condición | Obtener precios mayores o iguales a 100 |
| `reduce()` | Combina elementos en un solo valor | Sumar todos los precios |

**Para recordarlo:** `map` transforma, `filter` selecciona y `reduce` acumula.
