# 1. Operaciones aritméticas con Series

## ¿Qué es una Series?

Una Series es una estructura unidimensional de Pandas. Podemos imaginarla como una sola columna de una tabla, acompañada de un índice.

Una operación vectorizada permite aplicar una operación a todos los elementos de una Series sin escribir un ciclo `for` para recorrerlos uno por uno.


### 1.1 Seleccionar una Series


```python
precios = df["precio"]

print(type(precios))
display(precios.head())
```

### 1.2 Operaciones aritméticas

| Operación | Ejemplo |
|---|---|
| Suma | `precios + 20` |
| Resta | `precios - 20` |
| Multiplicación | `precios * 1.10` |
| División | `precios / 2` |
| Potencia | `precios ** 2` |

Cuando operamos una Series con un número (escalar), Pandas aplica la operación a cada elemento.


```python
# Aumentar cada precio un 10 %
aumento = precios * 1.10

# Aplicar un descuento del 15 %
descuento = precios * 0.85

# Sumar 20 pesos a cada precio
precio_mas_20 = precios + 20

# Calcular el cuadrado de cada precio
precio_al_cuadrado = precios ** 2

display(pd.DataFrame({
    "precio_original": precios.head(),
    "aumento_10": aumento.head(),
    "descuento_15": descuento.head(),
    "precio_mas_20": precio_mas_20.head()
}))
```

### 1.3 Operar dos Series

Pandas alinea las Series mediante sus índices. Si los índices coinciden, opera los valores correspondientes. Si no coinciden, el resultado puede contener `NaN` en los índices que no tengan pareja.



```python
serie1 = pd.Series([1, 2, 3], index=["A", "B", "C"])
serie2 = pd.Series([10, 20, 30], index=["A", "B", "C"])

print("Suma:")
display(serie1 + serie2)

print("Resta:")
display(serie2 - serie1)

print("Multiplicación:")
display(serie1 * serie2)

print("División:")
display(serie2 / serie1)
```

# 2. Operaciones aritméticas con DataFrames

Un DataFrame es una estructura bidimensional organizada en filas y columnas. Las operaciones aritméticas entre DataFrames se alinean por las etiquetas de filas y columnas, no solamente por la posición.

### 2.1 Crear dos DataFrames pequeños


```python
datos1 = {
    "enero": [100, 200, 300],
    "febrero": [150, 250, 350]
}

datos2 = {
    "enero": [50, 100, 150],
    "febrero": [50, 50, 50]
}

df1 = pd.DataFrame(datos1)
df2 = pd.DataFrame(datos2)

print("DataFrame 1")
display(df1)

print("DataFrame 2")
display(df2)

print("Suma")
display(df1 + df2)
```

### 2.2 Operaciones disponibles

Podemos utilizar los mismos operadores que vimos con Series:

- Suma: `df1 + df2`
- Resta: `df1 - df2`
- Multiplicación: `df1 * df2`
- División: `df1 / df2`
- Potencia: `df1 ** 2`
- Operación con escalar: `df1 * 1.10`

**Atención:** si los nombres de las columnas o los índices no coinciden, Pandas alinea por etiquetas y puede producir valores faltantes.


```python
print("Resta")
display(df1 - df2)

print("Multiplicación")
display(df1 * df2)

print("División")
display(df1 / df2)

print("Aumento del 10 %")
display(df1 * 1.10)
```

# 3. Funciones de agregación

Una agregación resume muchos valores en uno o varios resultados. Por ejemplo, en lugar de mostrar los 3,000 precios, podemos calcular el promedio, el mínimo o el máximo.

| Método | ¿Qué calcula? |
|---|---|
| `sum()` | Suma |
| `mean()` | Promedio |
| `median()` | Mediana |
| `min()` | Mínimo |
| `max()` | Máximo |
| `count()` | Cantidad de valores no nulos |
| `std()` | Desviación estándar |



```python
print("Suma de precios:", df["precio"].sum())
print("Precio promedio:", df["precio"].mean())
print("Mediana:", df["precio"].median())
print("Precio mínimo:", df["precio"].min())
print("Precio máximo:", df["precio"].max())
print("Cantidad de precios:", df["precio"].count())
print("Desviación estándar:", df["precio"].std())
```

### 3.1 Agrupar información con `groupby()`

`groupby()` permite dividir los registros en grupos según una columna y después calcular una agregación para cada grupo.

**Pregunta:** ¿Cuál es el precio promedio de los productos registrados en cada estado?


```python
precios_por_estado = df.groupby("estado")["precio"].mean()

display(precios_por_estado)
```

### 3.2 Varias agregaciones con `agg()`

Podemos solicitar varias estadísticas en una sola operación.


```python
resumen_estados = df.groupby("estado")["precio"].agg(
    ["count", "mean", "min", "max", "median"]
)

display(resumen_estados.head(10))
```

# 4. Limpieza de datos

La limpieza de datos consiste en identificar y tratar errores, inconsistencias, valores faltantes, duplicados y formatos incorrectos.

La calidad de los datos influye directamente en la calidad del análisis.

**Importante:** limpiar no significa borrar automáticamente todo lo que parece extraño. Primero debemos investigar y decidir qué tratamiento tiene sentido.


## 4.1 Identificar valores faltantes

`isnull()` o `isna()` identifica valores faltantes. Al combinarlo con `sum()` podemos contar cuántos hay por columna.


```python
# Contar valores faltantes por columna
faltantes = df.isnull().sum()
display(faltantes)

# Porcentaje de valores faltantes por columna
porcentaje_faltantes = df.isnull().mean() * 100
display(porcentaje_faltantes)
```

**Observación de nuestro archivo:** en la revisión inicial, las 16 columnas aparecen sin valores faltantes. Esto es un diagnóstico del archivo recibido, no significa que todos los CSV estén siempre completos.

Para practicar `dropna()` y `fillna()` crearemos una copia pequeña de ejemplo y simularemos un valor faltante.


```python
ejemplo_nulos = pd.DataFrame({
    "producto": ["Arroz", "Leche", "Pan"],
    "precio": [30.0, np.nan, 25.0]
})

display(ejemplo_nulos)

print("Valores faltantes:")
display(ejemplo_nulos.isna().sum())
```

### 4.2 Eliminar o imputar valores faltantes

- `dropna()` elimina filas que contienen valores faltantes.
- `dropna(axis=1)` elimina columnas que contienen valores faltantes.
- `fillna()` sustituye valores faltantes por otro valor.

La decisión depende del contexto y de la cantidad e importancia de los datos faltantes.


```python
# Eliminar filas con al menos un valor faltante
sin_nulos = ejemplo_nulos.dropna()
display(sin_nulos)

# Sustituir el precio faltante por la media
media_precio = ejemplo_nulos["precio"].mean()

rellenado = ejemplo_nulos.copy()
rellenado["precio"] = rellenado["precio"].fillna(media_precio)

display(rellenado)
```

### 4.3 Identificar y eliminar duplicados

`duplicated()` identifica registros repetidos. `drop_duplicates()` permite eliminarlos.

Antes de eliminar duplicados, debemos definir qué significa que dos registros sean iguales para nuestro análisis.


```python
# Contar filas completamente duplicadas
print("Duplicados exactos:", df.duplicated().sum())

# Revisar duplicados considerando algunas columnas
duplicados_producto_sucursal = df.duplicated(
    subset=["producto", "nombre_comercial", "municipio"]
)

print("Coincidencias en producto, sucursal y municipio:",
      duplicados_producto_sucursal.sum())
```

**Nota:** dos registros con el mismo producto, sucursal y municipio pueden ser legítimos si corresponden a fechas, presentaciones u otras condiciones diferentes. No los eliminaremos automáticamente.

Si confirmamos que existen filas completamente duplicadas y que deben retirarse, podemos trabajar sobre una copia:


```python
df_sin_duplicados = df.drop_duplicates().copy()

print("Filas originales:", len(df))
print("Filas después de quitar duplicados exactos:", len(df_sin_duplicados))
```

### 4.4 Detectar posibles valores atípicos con IQR

Un valor atípico es una observación que se aleja considerablemente del comportamiento general de los datos. Un valor alto no necesariamente es un error: puede corresponder a un producto costoso.

El rango intercuartílico se calcula así:

- Q1: percentil 25.
- Q3: percentil 75.
- IQR: Q3 menos Q1.
- Límite inferior: Q1 − 1.5 × IQR.
- Límite superior: Q3 + 1.5 × IQR.


```python
q1 = df["precio"].quantile(0.25)
q3 = df["precio"].quantile(0.75)

iqr = q3 - q1

limite_inferior = q1 - 1.5 * iqr
limite_superior = q3 + 1.5 * iqr

atipicos = df[
    (df["precio"] < limite_inferior) |
    (df["precio"] > limite_superior)
]

print("Q1:", q1)
print("Q3:", q3)
print("IQR:", iqr)
print("Límite inferior:", limite_inferior)
print("Límite superior:", limite_superior)
print("Posibles valores atípicos:", len(atipicos))

display(atipicos[["producto", "categoria", "precio"]].head(10))
```

### 4.5 Corregir formatos

Las fechas cargadas desde un CSV pueden llegar como texto (`object`). Para trabajar con ellas como fechas, podemos convertirlas con `pd.to_datetime()`.

También podemos limpiar espacios en textos con `.str.strip()` y transformar textos a minúsculas con `.str.lower()`.


```python
# Revisar el tipo de dato original
print(df["fecha_registro"].dtype)

# Convertir la fecha al tipo datetime
df_limpieza = df.copy()

df_limpieza["fecha_registro"] = pd.to_datetime(
    df_limpieza["fecha_registro"],
    format="%Y/%m/%d",
    errors="coerce"
)

print(df_limpieza["fecha_registro"].dtype)

# Ejemplos de limpieza de texto
df_limpieza["producto"] = df_limpieza["producto"].str.strip()
df_limpieza["estado"] = df_limpieza["estado"].str.strip()

display(df_limpieza[["producto", "fecha_registro", "estado"]].head())
```

# 5. Manipulación, reindexado y renombrado de columnas

Pandas permite adaptar la estructura del DataFrame a nuestras necesidades.

## 5.1 Agregar una nueva columna

Podemos crear una columna mediante una asignación. La operación se aplica a todos los registros.


```python
df_transformado = df.copy()

df_transformado["precio_con_iva"] = df_transformado["precio"] * 1.16

display(df_transformado[["producto", "precio", "precio_con_iva"]].head())
```

También podemos crear una columna categórica usando una condición con `.loc`.


```python
df_transformado["tipo_precio"] = "Hasta 100"

df_transformado.loc[
    df_transformado["precio"] > 100,
    "tipo_precio"
] = "Mayor a 100"

display(df_transformado[["producto", "precio", "tipo_precio"]].head(10))
```

## 5.2 Eliminar una columna

`drop(columns=[...])` elimina las columnas indicadas y devuelve un nuevo DataFrame, a menos que se solicite modificar el objeto original.


```python
df_sin_origen = df.drop(columns=["archivo_origen"])

print(df_sin_origen.columns.tolist())
```

## 5.3 Renombrar columnas

`rename()` permite cambiar uno o varios nombres de columnas.


```python
df_renombrado = df.rename(columns={
    "producto": "nombre_producto",
    "precio": "precio_mxn",
    "estado": "entidad"
})

print(df_renombrado.columns.tolist())
```

## 5.4 Reordenar columnas con `reindex()`

Podemos indicar el orden en que queremos visualizar las columnas. Las columnas que no incluyamos no aparecerán en el resultado.


```python
columnas_seleccionadas = [
    "producto",
    "marca",
    "categoria",
    "precio",
    "estado",
    "municipio"
]

df_reordenado = df.reindex(columns=columnas_seleccionadas)

display(df_reordenado.head())
```

# 6. Ejercicio integrador: análisis de precios por estado

## Contexto

Una empresa quiere analizar los precios de diferentes productos registrados en establecimientos comerciales de México. Nos solicita preparar la información y generar un resumen que permita comparar los precios entre estados.

### Instrucciones

1. Carga el CSV.
2. Muestra las primeras 10 filas.
3. Revisa dimensiones, columnas y tipos de datos.
4. Calcula precio promedio, mínimo y máximo.
5. Crea `precio_con_iva` aplicando 16 %.
6. Crea `precio_descuento` aplicando 15 %.
7. Identifica registros duplicados.
8. Revisa valores faltantes.
9. Convierte `fecha_registro` a tipo fecha.
10. Calcula el precio promedio por estado.
11. Renombra `producto` como `nombre_producto` y `precio` como `precio_mxn`.
12. Selecciona y reordena las columnas para crear un DataFrame final.

**Entrega esperada:** un DataFrame preparado para realizar análisis posteriores y una tabla resumen con el precio promedio por estado.


```python
# 1. Trabaja sobre una copia
df_final = df.copy()

# 2. Revisa los datos
display(df_final.head(10))
print(df_final.shape)
df_final.info()

# 3. Calcula estadísticas generales
print("Promedio:", df_final["precio"].mean())
print("Mínimo:", df_final["precio"].min())
print("Máximo:", df_final["precio"].max())

# 4. Crea columnas calculadas
df_final["precio_con_iva"] = df_final["precio"] * 1.16
df_final["precio_descuento"] = df_final["precio"] * 0.85

# 5. Revisa faltantes y duplicados
print(df_final.isna().sum())
print("Duplicados exactos:", df_final.duplicated().sum())

# 6. Convierte la fecha
df_final["fecha_registro"] = pd.to_datetime(
    df_final["fecha_registro"],
    format="%Y/%m/%d",
    errors="coerce"
)

# 7. Resumen por estado
resumen_estado = df_final.groupby("estado")["precio"].mean().sort_values(
    ascending=False
)
display(resumen_estado)

# 8. Renombra columnas
df_final = df_final.rename(columns={
    "producto": "nombre_producto",
    "precio": "precio_mxn"
})

# 9. Selecciona y ordena columnas
columnas_finales = [
    "nombre_producto",
    "marca",
    "categoria",
    "precio_mxn",
    "precio_con_iva",
    "precio_descuento",
    "fecha_registro",
    "estado",
    "municipio"
]

df_final = df_final.reindex(columns=columnas_finales)

display(df_final.head())
```
