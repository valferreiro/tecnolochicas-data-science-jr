Desarrolla un programa en Python que permita ingresar el nombre del empleado, su salario diario y la cantidad de horas extras trabajadas en una semana. El programa arrojará cuánto se le debe pagar al empleado por las horas extras. El programa también debe realizar el cálculo automático del pago correspondiente por las horas extras del empleado, siguiendo estos pasos: 
- Solicitará al usuario el nombre del trabajador.
- Pedirá al usuario que ingrese el salario diario del empleado.
- Solicitará al usuario ingresar las horas extras trabajadas por el empleado en una semana.
  
A continuación, procederá a calcular el total del pago por las horas extras. Para ello, el programa: 
- Dividirá el salario diario entre 8 para determinar la tarifa por hora trabajada. 
- Si las horas extras trabajadas son 9 o menos, se pagarán al doble. 
- Si las horas extras trabajadas superan las 9, se pagarán las primeras 9 horas al doble y las horas restantes al triple. 

Finalmente, el programa mostrará un mensaje que incluirá el nombre del empleado y el total de su pago por las horas extras.

```python
print("-----------------------------------------------")
print("Bienvenidx al Sistema de Pagos por Horas Extras")
print("-----------------------------------------------")

nombre_empleado = input("\nIngresa el nombre del empleado:")
salario_diario = float(input("Ingresa el salario diario del empleado {}:".format(nombre_empleado)))
horas_extras = int(input("Ingresa la cantidad de horas extras que el empleado {} realizó:".format(nombre_empleado)))

# Calcular el pago por hora
pago_hora = salario_diario / 8

# Calcular el pago por horas extras
if horas_extras <= 9:
    pago_extra = horas_extras * pago_hora * 2

else:
    pago_doble = 9 * pago_hora * 2
    pago_triple = (horas_extras - 9) * pago_hora * 3
    pago_extra = pago_doble + pago_triple

print("\nEl pago de las {} horas extras del empleado {}, es de: ${:,.2f}".format(horas_extras, nombre_empleado, pago_extra))
```
